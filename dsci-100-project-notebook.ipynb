#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

```r
marchmad_4.1 <- march_mad_data |>
select(conf_wins_pct, reg_wins_pct) |>  #selecting numerical columns for this step 

pivot_longer( cols = everything(),
              names_to = "variable",
              values_to = "value") |> #making it longer and easier to read so min mean and max for each column dont all become columns

group_by(variable) |> #so we can see values for each and theyre clearly labelled

summarize( min = min(value, na.rm = TRUE),
           mean = mean(value, na.rm = TRUE),
           max = max(value, na.rm = TRUE)) #finding the min mean and max for each, dropping NA's 
           
marchmad_4.1
```

#### Task 4.2: Determine the amount of missingness (2 points)

- Use your `dplyr` knowledge to create **one** table that calculates the number of missing values for each column in the data set (both numeric and categorical).
- Show the outputted table.

```r
marchmad4.2 <- march_mad_data |>

select(conf_wins_pct, reg_wins_pct, tourney_finish) |> #while i have year and school for visual sake they are irrelevant so taken out
mutate(conf_wins_pct = as_factor(conf_wins_pct),
       reg_wins_pct = as_factor(reg_wins_pct)) |> #since what kind of values they are doesnt matter here (just finding missing) i can turn the 2 numericals into factors for simplicity

pivot_longer( cols = everything(),
              names_to = "column",
              values_to = "count") |> #similar step as before but naming values as "count" since I'm finding # of missing values now

group_by(column) |>

summarize(count = sum(is.na(count)))
marchmad4.2
```

I will first create a plot that puts regional wins against conference wins, to see if they have a relationship. If there's little to no relationship, we can begin to determine if a team is truly good, or if the competition in their conference was simply weak in comaprison. In comparison, if the two are closely related (for example a school has a high or low win percentage for both), we can see a trend potentially demonstrating a teams performance, since a strong relationship between the two would tell us that the school consistently wins or loses, which will then help to determine if/how win rates are related to placements. 

```r
winscompair <- march_mad_data |>

select(conf_wins_pct, reg_wins_pct) |> #selecting desired variables for plot
filter(conf_wins_pct != "NA", reg_wins_pct != "NA") |> #removing NA values

ggplot( aes( x = conf_wins_pct, y = reg_wins_pct)) +
geom_point(alpha = 0.2, size = 0.7, color = "purple") + #hard to avoid overplotting here unless i were to restrict years + im choosing purple to not go insane
xlab("Confrence Win %") +
ylab("Regionals Win %") +
ggtitle("Confrence vs Regional Win Percentages") +
theme(text = element_text(size = 15))

winscompair #resulting graph shows a positive, moderately strong linear relationship between the two variables
```
